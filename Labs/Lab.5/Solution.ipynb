Matrix assignment works in 2 ways:
If M_1 and M_2 are matrix instances M_1=M_2 sets the values of M_1 to those of M_2, if they are the same size. Error otherwise.
In example above M_2 can be a list of lists of correct size.

```python
class Matrix:

    def __init__(self, values, cols=None):
    #There are two ways to create a matrix. If the user gives two integers, I create a matrix using nested loops. If the user gives a list of lists, I copy the values into self.data, converting them to floats.

        if isinstance(values, int) and isinstance(cols, int):

            self.rows = values
            self.cols = cols
            self.data = []

            for i in range(self.rows):
                row = []
                for j in range(self.cols):
                    row.append(0.0)
                self.data.append(row)
        elif isinstance(values, list):

            if len(values) == 0:
                raise ValueError("Matrix cannot be empty")

            first_row_length = len(values[0])

            for row in values:
                if len(row) != first_row_length:
                    raise ValueError("All rows must have same length")

            self.rows = len(values)
            self.cols = first_row_length
            self.data = []

            for row in values:
                new_row = []
                for item in row:
                    new_row.append(float(item))
                self.data.append(new_row)

        else:
            raise TypeError("Invalid input ")

    # allows M[i][j] and M[i,j]
    def __getitem__(self, key):

        # if user writes M[i,j]
        if isinstance(key, tuple):
            row_index = key[0]
            col_index = key[1]
            return self.data[row_index][col_index]

        # if user writes M[i]
        return self.data[key]

    # allows setting values like M[i,j] = value
    def __setitem__(self, key, value):

        if isinstance(key, tuple):
            row_index = key[0]
            col_index = key[1]
            self.data[row_index][col_index] = float(value)
        else:
            self.data[key] = value

    def shape(self):
        return (self.rows, self.cols)
    
    #Python does not allow overloading the = operator. So I created an assign() method that copies values from another matrix (or list of lists) into the current matrix, as long as the sizes match.

    def assign(self, other):

        if isinstance(other, Matrix):

            if self.shape() != other.shape():
                raise ValueError("Matrix size must match")

            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = other.data[i][j]

   
        elif isinstance(other, list):

            if len(other) != self.rows:
                raise ValueError("Matrix sizes must match")

            for row in other:
                if len(row) != self.cols:
                    raise ValueError("Matrix sizes must match")

            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = float(other[i][j])

        else:
            raise TypeError("Invalid assignment type")
```

```python
class Matrix:

    def __init__(self, values, cols=None):
        if isinstance(values, int) and isinstance(cols, int):

            self.rows = values
            self.cols = cols
            self.data = []

            for i in range(self.rows):
                row = []
                for j in range(self.cols):
                    row.append(0.0)
                self.data.append(row)

        elif isinstance(values, list):

            if len(values) == 0:
                raise ValueError("Matrix cannot be empty")

            first_row_length = len(values[0])

            for row in values:
                if len(row) != first_row_length:
                    raise ValueError("All rows must have same length")

            self.rows = len(values)
            self.cols = first_row_length
            self.data = []

            for row in values:
                new_row = []
                for item in row:
                    new_row.append(float(item))
                self.data.append(new_row)

        else:
            raise TypeError("Invalid input")

    def __getitem__(self, key):

        if isinstance(key, tuple):
            row_index = key[0]
            col_index = key[1]
            return self.data[row_index][col_index]

        return self.data[key]

    def __setitem__(self, key, value):

        if isinstance(key, tuple):
            row_index = key[0]
            col_index = key[1]
            self.data[row_index][col_index] = float(value)
        else:
            self.data[key] = value

    def shape(self):
        return (self.rows, self.cols)

    # Python does not allow overloading =
    # so assign() copies values if sizes match
    def assign(self, other):

        if isinstance(other, Matrix):

            if self.shape() != other.shape():
                raise ValueError("Matrix size must match")

            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = other.data[i][j]

        elif isinstance(other, list):

            if len(other) != self.rows:
                raise ValueError("Matrix sizes must match")

            for row in other:
                if len(row) != self.cols:
                    raise ValueError("Matrix sizes must match")

            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = float(other[i][j])

        else:
            raise TypeError("Invalid assignment type")

    def to_list(self):
        return self.data

    def transpose(self):

        new_data = []

        for j in range(self.cols):

            new_row = []

            for i in range(self.rows):
                new_row.append(self.data[i][j])

            new_data.append(new_row)

        return Matrix(new_data)

   # I return [self.data[n]] so that it stays a matrix instead of just a list.
    def row(self, n):
        return Matrix([self.data[n]])

    # returns nth column as matrix
    def column(self, n):

        new_data = []

        for i in range(self.rows):
            new_data.append([self.data[i][n]])

        return Matrix(new_data)

    def block(self, n0, n1, m0, m1): #I use nested loops to extract only the rows and columns in the given range

        new_data = []

        for i in range(n0, n1):

            new_row = []

            for j in range(m0, m1):
                new_row.append(self.data[i][j])

            new_data.append(new_row)

        return Matrix(new_data)
```

Write functions that create special matrices (note these are standalone functions, not member functions of your matrix class):
constant(n,m,c): returns a n by m matrix filled with floats of value c.
zeros(n,m) and ones(n,m): return n by m matrices filled with floats of value 0 and 1, respectively.
eye(n): returns the n by n identity matrix.

```python
def constant(n, m, c): #I use nested loops to build a list of lists filled with the given value and return a Matrix object.
    data = []
    for i in range(n):

        row = []

        for j in range(m):
            row.append(float(c))

        data.append(row)

    return Matrix(data)

def zeros(n, m):
    return constant(n, m, 0) #I reused constant() to avoid repeating code.

def ones(n, m):
    return constant(n, m, 1)

def eye(n): #I check if row index equals column index. If so, put 1.0, otherwise 0.0.

    data = []

    for i in range(n):

        row = []

        for j in range(n):

            if i == j:
                row.append(1.0)
            else:
                row.append(0.0)

        data.append(row)

    return Matrix(data)
```

```python
print("Constant matrix:")
A = constant(2, 3, 5)
print(A.to_list())

print("\nZeros matrix:")
B = zeros(2, 2)
print(B.to_list())

print("\nOnes matrix:")
C = ones(2, 2)
print(C.to_list())

print("\nIdentity matrix:")
print(eye(3).to_list())
```
