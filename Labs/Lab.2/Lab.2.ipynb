*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
def check_winner(board):
    n = len(board)
    
    for player in [1, 2]:
        for row in board:
            if all(cell == player for cell in row):
                return player
        for col in range(n):
            if all(board[row][col] == player for row in range(n)):
                return player
        if all(board[i][i] == player for i in range(n)):
            return player
        if all(board[i][n-1-i] == player for i in range(n)):
            return player
    
    if any(board[r][c] == 0 for r in range(n) for c in range(n)):
        return -1
    
    return 0
````

````python
incomplete = [[1, 0, 0],
              [0, 2, 0],
              [0, 0, 0]]
assert check_winner(incomplete) == -1

draw = [[1, 2, 1],
        [1, 1, 2],
        [2, 1, 2]]
assert check_winner(draw) == 0

x_wins_row = [[1, 1, 1],
              [2, 2, 0],
              [0, 0, 0]]
assert check_winner(x_wins_row) == 1

o_wins_col = [[1, 2, 1],
              [1, 2, 0],
              [0, 2, 0]]
assert check_winner(o_wins_col) == 2

x_wins_diag = [[1, 2, 0],
               [2, 1, 0],
               [0, 2, 1]]
assert check_winner(x_wins_diag) == 1

o_wins_anti = [[1, 1, 2],
               [1, 2, 0],
               [2, 0, 0]]
assert check_winner(o_wins_anti) == 2

print("All tests passed ✓")
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
def place_mark(board, player, row, col):
    if board[row][col] != 0:
        return False
    board[row][col] = player
    return True
````

````python
board = [[0, 0, 0],
         [0, 0, 0],
         [0, 0, 0]]

assert place_mark(board, 1, 0, 0) == True
assert board[0][0] == 1

assert place_mark(board, 2, 1, 1) == True
assert board[1][1] == 2

assert place_mark(board, 1, 0, 0) == False
assert board[0][0] == 1

assert place_mark(board, 2, 1, 1) == False
assert board[1][1] == 2

assert place_mark(board, 1, 2, 2) == True
assert board[2][2] == 1

print("All tests passed ✓")
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
def place_mark_str(board, player, location):
    row_labels = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    row = row_labels.index(location[0].upper())
    col = int(location[1:]) - 1
    return place_mark(board, player, row, col)
````

````python
board = create_board(3)

assert place_mark_str(board, 1, "A1") == True
assert board[0][0] == 1

assert place_mark_str(board, 2, "B2") == True
assert board[1][1] == 2

assert place_mark_str(board, 1, "A1") == False
assert board[0][0] == 1

assert place_mark_str(board, 2, "c3") == True
assert board[2][2] == 2

assert place_mark_str(board, 1, "B2") == False

print("All tests passed ✓")
````

Keep asking for input until the player enters a valid location that results in a valid move.


````python
def get_player_move(board, player):
    while True:
        location = input(f"Player {player}, enter your move (e.g. A1): ")
        try:
            if place_mark_str(board, player, location):
                return
        except (ValueError, IndexError):
            pass
        print("Invalid move, try again.")
````

- draw the board,

- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
def play_game(n=3):
    board = create_board(n)
    player = 1

    while True:
        draw_board(board)
        get_player_move(board, player)
        result = check_winner(board)

        if result == 1:
            draw_board(board)
            print("Player 1 (X) wins!")
            return
        elif result == 2:
            draw_board(board)
            print("Player 2 (O) wins!")
            return
        elif result == 0:
            draw_board(board)
            print("It's a draw!")
            return

        player = 2 if player == 1 else 1
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
def get_computer_move(board, computer, human):
    n = len(board)

    def try_win(player):
        for r in range(n):
            for c in range(n):
                if board[r][c] == 0:
                    board[r][c] = player
                    if check_winner(board) == player:
                        board[r][c] = 0
                        return (r, c)
                    board[r][c] = 0
        return None

    move = try_win(computer)
    if move:
        place_mark(board, computer, *move)
        return

    move = try_win(human)
    if move:
        place_mark(board, computer, *move)
        return

    center = n // 2
    if board[center][center] == 0:
        place_mark(board, computer, center, center)
        return

    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                place_mark(board, computer, r, c)
                return

def play_vs_computer(n=3):
    board = create_board(n)
    human = 1
    computer = 2

    while True:
        draw_board(board)
        get_player_move(board, human)
        result = check_winner(board)
        if result == human:
            draw_board(board)
            print("You win!")
            return
        if result == 0:
            draw_board(board)
            print("It's a draw!")
            return

        get_computer_move(board, computer, human)
        result = check_winner(board)
        if result == computer:
            draw_board(board)
            print("Computer wins!")
            return
        if result == 0:
            draw_board(board)
            print("It's a draw!")
            return
    
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
def minimax(board, player, opponent, depth, maximizing):
    result = check_winner(board)
    if result == player:
        return 10
    if result == opponent:
        return -10
    if result == 0:
        return 0
    if depth == 0:
        return 0

    n = len(board)
    if maximizing:
        best = -float('inf')
        for r in range(n):
            for c in range(n):
                if board[r][c] == 0:
                    board[r][c] = player
                    best = max(best, minimax(board, player, opponent, depth-1, False))
                    board[r][c] = 0
        return best
    else:
        best = float('inf')
        for r in range(n):
            for c in range(n):
                if board[r][c] == 0:
                    board[r][c] = opponent
                    best = min(best, minimax(board, player, opponent, depth-1, True))
                    board[r][c] = 0
        return best

def get_computer_move_minimax(board, computer, human, depth=6):
    n = len(board)
    best_score = -float('inf')
    best_move = None

    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                board[r][c] = computer
                score = minimax(board, computer, human, depth-1, False)
                board[r][c] = 0
                if score > best_score:
                    best_score = score
                    best_move = (r, c)

    if best_move:
        place_mark(board, computer, *best_move)

def play_vs_minimax(n=3, depth=6):
    board = create_board(n)
    human = 1
    computer = 2

    while True:
        draw_board(board)
        get_player_move(board, human)
        result = check_winner(board)
        if result == human:
            draw_board(board)
            print("You win!")
            return
        if result == 0:
            draw_board(board)
            print("It's a draw!")
            return

        get_computer_move_minimax(board, computer, human, depth)
        result = check_winner(board)
        if result == computer:
            draw_board(board)
            print("Computer wins!")
            return
        if result == 0:
            draw_board(board)
            print("It's a draw!")
            return
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
def play_computer_vs_computer(n, depth_p1, depth_p2, games=10):
    results = {1: 0, 2: 0, 0: 0}

    for _ in range(games):
        board = create_board(n)
        player, opponent = 1, 2

        while True:
            if player == 1:
                get_computer_move_minimax(board, player, opponent, depth_p1)
            else:
                get_computer_move_minimax(board, player, opponent, depth_p2)

            result = check_winner(board)
            if result != -1:
                results[result] += 1
                break

            player, opponent = opponent, player

    print(f"\n{n}x{n} grid | depth_p1={depth_p1} vs depth_p2={depth_p2} | {games} games")
    print(f"  Player 1 wins: {results[1]}")
    print(f"  Player 2 wins: {results[2]}")
    print(f"  Draws:         {results[0]}")
    smarter_wins = results[2]
    print(f"  Smarter (P2, depth={depth_p2}) win rate: {smarter_wins/games*100:.1f}%")
````

````python
play_computer_vs_computer(n=3, depth_p1=1, depth_p2=6, games=10)
play_computer_vs_computer(n=4, depth_p1=1, depth_p2=3, games=10)
play_computer_vs_computer(n=5, depth_p1=1, depth_p2=2, games=10)
````
